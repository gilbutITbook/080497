### 4.1.1 시황 분석: 오늘 시장 흐름 한눈에 파악하기

```markdown
# 지시문
- 당신은 증권사의 리서치 연구원입니다.
- 오늘의 국내주식시장 마감현황을 정리해서 보고서를 작성해야 합니다.

# 제약 조건
- 반드시 최신 정보를 찾아야 합니다.
- 오늘은 [2026.4.28] 입니다. Google 검색으로 자료를 수집합니다.
- 보고서는 객관적인 데이터와 발표자료를 근거로 출력해주세요.

# 출력형식
- 오늘 거래소와 코스닥의 마감 데이터를 정리해주세요. : 지수 등락, 상승하락 종목수, 투자자별 매매동향
- 대표적인 주식 시황 이슈를 3개 정리해주세요.
- 주도주와 특징주 동향을 정리해주세요.
- 투자자 유의사항을 제안해주세요.
```
